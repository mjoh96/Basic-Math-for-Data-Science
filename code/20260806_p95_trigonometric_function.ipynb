```python
# Ex 2-19
import numpy as np
import matplotlib.pyplot as plt

#y=sin2(x+1)
x=np.linspace(-1,np.pi-1,501)
y=np.sin(2*(x+1))
plt.plot(x,y,label=r'$y=sin2(x+1)$')
plt.legend()
plt.show()

#y=-3cosx+1
x=np.linspace(0,2*np.pi,501)
y=-3*np.cos(x)+1
plt.plot(x,y,label=r'$y=-3cosx+1$')
plt.legend()
plt.show()
```
