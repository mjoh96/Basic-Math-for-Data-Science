```python
#Ch 3 프로그래밍 실습3
import matplotlib.pyplot as plt

#주요 함수
a1=109
an=a1
anp1=an

n_list=[]
n_list.append(1)
an_list=[]
an_list.append(a1)

for n in range(1,100000):
    if (an%2==0):
        anp1=an/2
    else:
        anp1=3*an+1
    n_list.append(n+1)
    an_list.append(anp1)

    print(anp1)
    if anp1==1:
        print("n=",n+1)
        break

    an=anp1

plt.plot(n_list,an_list,'-o')
plt.show()
    
```
