```python
#Ch1 프로그래밍 실습2
import numpy as np
import matplotlib.pyplot as plt

#f(x)
def f(x):
    if x>=-1 and x<1:
        return -x**2 +2
    if x>=1 and x<=3:
        return x**2-4*x+3
    if x>3 and x<=5:
        return -x**2+8*x-14

#x, fx 초기화
x=np.linspace(-1,5,601)
fx=np.zeros_like(x)

#f(x)의 값 계산
for i in range (0,601):
    fx[i]=f(x[i])

#y=f(x)의 그래프
plt.plot(x,fx,'.')
plt.show()
```
