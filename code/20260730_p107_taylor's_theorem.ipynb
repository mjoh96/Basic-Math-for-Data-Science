```python
# Ch2 프로그래밍 실습3
import numpy as np
import matplotlib.pyplot as plt

def f(x):
    return np.sin(x)
def T1(x):
    return x
def T2(x):
    return T1(x)-(1/6)*x**3
def T3(x):
    return T2(x)+(1/120)*x**5
def T4(x):
    return T3(x)-(1/5040)*x**7

#주요 함수
x=np.linspace(-np.pi,np.pi,1001)
y=f(x)
y1=T1(x)
y2=T2(x)
y3=T3(x)
y4=T4(x)

graph, = plt.plot(x,y)
graph1, = plt.plot(x,y1)
graph2, = plt.plot(x,y2)
graph3, = plt.plot(x,y3)
graph4, = plt.plot(x,y4)

plt.legend(handles=(graph,graph1,graph2,graph3,graph4),
labels=(r'$y=sin(x)$',r'$y=T_1(x)$',r'$y=T_2(x)$',r'$y=T_3(x)$',r'$y=T_4(x)$'))
```
