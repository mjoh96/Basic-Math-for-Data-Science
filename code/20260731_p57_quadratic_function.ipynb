```python
# Ex 2-2
import numpy as np
import matplotlib.pyplot as plt

# (a) y=0.5(x-1)^2+1
x=np.linspace(-3,5,801)
y=0.5*(x-1)**2+1
plt.plot(x,y,label=r'$y=0.5(x-1)^2+1$')
plt.legend()
plt.show()

# (b) y=-(x+1)^2-2
x=np.linspace(-5,3,801)
y=-(x+1)**2-2
plt.plot(x,y,label='$y=-(x+1)^2-2$')
plt.legend()
plt.show
```
