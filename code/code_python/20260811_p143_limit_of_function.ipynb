```python
# Ex 3-21
import numpy as np
import matplotlib.pyplot as plt

# (a) f(x)=x^2-1
x=np.linspace(0,5,201)
fx=x**2-1
x1=3
fx1=x1**2-1
plt.plot(x,fx,'k',label=r'$\lim_{x \to 3}(x^2-1)$')
plt.plot(x1,fx1,'o')
plt.legend()
plt.show()

#(b) g(x)=(x^2-9)/(x-3)
x1=np.linspace(-6,-3-0.015,101)
x2=-3
x3=np.linspace(-3+0.015,-0,101)
gx1=(x1**2-9)/(x1+3)
gx2=-6
gx3=(x3**2-9)/(x3+3)
plt.plot(x2,gx2,'o',mfc='none')
plt.plot(x3,gx3,'k')
plt.plot(x1,gx1,'k',label=r'$\lim_{x \to -3}$$\frac{x^2-9}{x+3}$')
plt.legend()
plt.show()
```
