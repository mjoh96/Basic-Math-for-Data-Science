```python
import numpy as np
import matplotlib.pyplot as plt

#(a) y1=2^x, y2=2^(x+1)-3
x=np.linspace(-2,1,301)
y1=2**x
y2=2**(x+1)-3

#그래프 그리기
graph1, = plt.plot(x,y1)
graph2, = plt.plot(x,y2)
plt.legend(handles=(graph1,graph2), labels=(r'$y=2^x$',r'$y=2^{x+1}-3$'))
plt.show()

#(b) y1=3^(-x), y2=-3^(-x+1)
x=np.linspace(-2,1,301)
y1=3**(-x)
y2=-3**(-x+1)

#그래프 그리기
graph1, = plt.plot(x,y1)
graph2, = plt.plot(x,y2)
plt.legend(handles=(graph1,graph2),labels=(r'$y=3^{-x}$',r'$y=-3^{-x+1}$'))
plt.show()
```
