```python
#Ex 2-6
import numpy as np
import matplotlib.pyplot as plt

# (a) y=3/(x-1)+4
x1=np.linspace(-0.75,0.9,100)
x2=np.linspace(1.1,2.00,100)

y1=3/(x1-1)+4
y2=3/(x2-1)+4

plt.plot(x1,y1,label=r'$y=\frac{3}{x-1}+4$')
plt.plot(x2,y2,label=r'$y=\frac{3}{x-1}+4$')
plt.legend()
plt.show()
```

```python
#(b) y=-1/(3x+2)-2

x1=np.linspace(-2,-2/3-0.1,100)
x2=np.linspace(-2/3+0.1,0.5,100)

y1=-1/(3*x1+2)-2
y2=-1/(3*x2+2)-2

plt.plot(x1,y1,label=r'$y=\frac{-1}{3x+2}+2$')
plt.plot(x2,y2,label=r'$y=\frac{-1}{3x+2}+2$')
plt.legend()
plt.show()
```
