```python
#Ex 2-7
import numpy as np
import matplotlib.pyplot as plt

#y=(-4x+3)/(2x-3)
x1=np.linspace(0,1.5-0.1,1000)
x2=np.linspace(1.5+0.1,3,1000)

y1=(-4*x1+3)/(2*x1-3)
y2=(-4*x2+3)/(2*x2-3)

plt.plot(x1,y1,label=r'y=$\frac{-4x+3}{2x-3}$')
plt.plot(x2,y2,label=r'y=$\frac{-4x+3}{2x-3}$')
plt.legend()
plt.show()
```
